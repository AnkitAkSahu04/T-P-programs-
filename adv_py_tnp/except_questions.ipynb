```python
'''STUDENT GRADE SYSTEM'''
students = {}

def add_student():
    try:
        sid = input("Enter Student ID: ")
        if sid in students:
            raise Exception("Student ID already exists")

        grade = input("Enter Grade: ")

        if grade == "":
            raise ValueError("Grade cannot be empty")

        grade = float(grade)

        students[sid] = grade
        print("Student added successfully")

    except ValueError as e:
        print("Error:", e)

    except Exception as e:
        print("Error:", e)


def update_student():
    try:
        sid = input("Enter Student ID: ")

        if sid not in students:
            raise KeyError("Invalid Student ID")

        grade = float(input("Enter new grade: "))
        students[sid] = grade

        print("Grade updated")

    except ValueError:
        print("Grade must be a number")

    except KeyError as e:
        print(e)


def delete_student():
    try:
        sid = input("Enter Student ID: ")

        if sid not in students:
            raise KeyError("Student not found")

        del students[sid]
        print("Student deleted")

    except KeyError as e:
        print(e)


while True:
    print("\n1.Add 2.Update 3.Delete 4.Show 5.Exit")
    choice = input("Choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        update_student()

    elif choice == "3":
        delete_student()

    elif choice == "4":
        print(students)

    elif choice == "5":
        break
```

```python
'''CONTACT BOOK'''
contacts = {}

while True:
    try:
        print("\n1 Add 2 Search 3 Edit 4 Show 5 Exit")
        ch = input("Choice: ")

        if ch == "1":
            name = input("Enter name: ")
            phone = input("Enter phone: ")

            if name == "" or phone == "":
                raise ValueError("Fields cannot be empty")

            if not phone.isdigit():
                raise ValueError("Phone must contain digits only")

            if name in contacts:
                raise Exception("Duplicate contact")

            contacts[name] = phone
            print("Contact added")

        elif ch == "2":
            name = input("Search name: ")

            if name not in contacts:
                raise KeyError("Contact not found")

            print("Phone:", contacts[name])

        elif ch == "3":
            name = input("Edit name: ")

            if name not in contacts:
                raise KeyError("Contact not found")

            phone = input("New phone: ")
            contacts[name] = phone

        elif ch == "4":
            print(contacts)

        elif ch == "5":
            break

    except Exception as e:
        print("Error:", e)
```

```python
'''BANKING SYSTEM WITH TRANSACTION'''
accounts = {"101":5000, "102":3000}

try:
    sender = input("Sender account: ")
    receiver = input("Receiver account: ")
    amount = float(input("Amount: "))

    if sender not in accounts or receiver not in accounts:
        raise Exception("Invalid account number")

    if accounts[sender] < amount:
        raise Exception("Overdraft! Not enough balance")

    accounts[sender] -= amount
    accounts[receiver] += amount

    print("Transaction successful")

except ValueError:
    print("Invalid amount")

except Exception as e:
    print("Error:", e)

print(accounts)
```

```python
'''E-COMMERCE ORDER MANAGEMENT'''
products = {"laptop":5, "phone":3}
coupons = ["SAVE10"]

try:
    item = input("Enter product: ")

    if item not in products:
        raise Exception("Product not found")

    qty = int(input("Quantity: "))

    if qty > products[item]:
        raise Exception("Out of stock")

    coupon = input("Enter coupon: ")

    if coupon != "" and coupon not in coupons:
        raise Exception("Invalid coupon code")

    payment = input("Payment method (card/cash): ")

    if payment not in ["card","cash"]:
        raise Exception("Invalid payment method")

    products[item] -= qty
    print("Order successful")

except Exception as e:
    print("Error:", e)
```

```python
'''CUSTOM EXCEPTION FRAMEWORK'''
class OutOfStockError(Exception):
    pass

class InvalidProductIDError(Exception):
    pass

inventory = {"101":5, "102":2}

try:
    pid = input("Enter product id: ")

    if pid not in inventory:
        raise InvalidProductIDError("Product ID invalid")

    qty = int(input("Quantity: "))

    if qty > inventory[pid]:
        raise OutOfStockError("Not enough stock")

    inventory[pid] -= qty
    print("Purchase successful")

except InvalidProductIDError as e:
    print(e)

except OutOfStockError as e:
    print(e)
```

```python
'''FLIGHT SYSTEM'''
flights = {"AI101":5, "AI202":2}

try:
    flight = input("Enter flight number: ")

    if flight not in flights:
        raise Exception("Invalid flight number")

    seats = int(input("Seats required: "))

    if seats > flights[flight]:
        raise Exception("Seat not available")

    name = input("Passenger name: ")

    if name == "":
        raise Exception("Invalid passenger details")

    payment = input("Payment success? (yes/no): ")

    if payment != "yes":
        raise Exception("Payment failed")

    flights[flight] -= seats
    print("Flight booked successfully")

except Exception as e:
    print("Error:", e)
```
