```python
'''SUM OF DIGIT OF A NUMBER'''
n =6789
sum = 0
while n > 0:
    sum += n % 10
    n //= 10
print(sum)
```

```python
'''Remove duplicates from a list (without set).'''
lst = [1,4,3,3,4,4,15,4,6,8,8,9,10]
res = []
for i in lst:
    if i not in res:
        res.append(i)
print(res)
```

```python
'''Count frequency of elements in a list.'''
lst = [1,2,2,3,1,4,5,7]
freq = {}
for i in lst:
    freq[i] = freq.get(i, 0) + 1
print(freq)
```

```python
'''Find the first non-repeating character.'''
s = "aabcddeeffg"
for ch in s:
    if s.count(ch) == 1:
        print(ch)
        break
```

```python
'''Flatten a nested list.'''
lst = [[1,2],[3,4],[5,3],[4]]
flat = []
for sub in lst:
    for i in sub:
        flat.append(i)
print(flat)
```

```python
'''Find all pairs with a given sum.'''
lst = [1,2,3,4,5,6,7,8]
target = 7
for i in range(len(lst)):
    for j in range(i+1, len(lst)):
        if lst[i] + lst[j] == target:
            print(lst[i], lst[j])
```

```python
'''Print a right triangle pattern.'''
n = 9
for i in range(1, n+1):
    print("*" * i)
```

```python
'''Print a pyramid pattern.'''
n=10
for i in range (1,n+1):
  print(" "*(n-i)+"*"*(2*i-1))
```

```python
'''Print all prime numbers in a range.'''
for n in range(2, 50):
    for i in range(2, n):
        if n % i == 0:
            break
    else:
        print(n)
```

```python
'''ARMSTRONG NUMBER'''
n = int(input('Enter a number: '))
digits = str(n)
total = sum(int(i)**len(digits) for i in digits)

if total == n:
    print('It is an Armstrong number.')
else:
    print('It is not an Armstrong number.')
```
